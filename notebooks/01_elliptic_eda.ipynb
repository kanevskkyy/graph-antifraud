Verify installation

```python
import torch
import torch_geometric

print("PyTorch:", torch.__version__)
print("PyG:", torch_geometric.__version__)
print("GPU available:", torch.cuda.is_available())
```

Load dataset

```python
from torch_geometric.datasets import EllipticBitcoinDataset

dataset = EllipticBitcoinDataset(root="/content/elliptic")
data = dataset[0]

print(data)
```

Explore the graph

```python
import pandas as pd

print("Number of nodes:", data.num_nodes)
print("Number of edges:", data.edge_index.shape[1])
print("Number of features per node:", data.x.shape[1])

print("\nLabel distribution:")
print(pd.Series(data.y.numpy()).value_counts())
```

Inspect dataset masks

```python
for name in ["train_mask", "val_mask", "test_mask"]:
    if hasattr(data, name):
        print(name, getattr(data, name).sum().item())
```
